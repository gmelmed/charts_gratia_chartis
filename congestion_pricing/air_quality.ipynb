```python
sites_filepath = "data/raw/air_quality_nyc_sites.csv"
```

```python
sites = pd.read_csv(sites_filepath)
sites = sites[~sites['aqs_id_full'].isna()]

sites
```

```python
sites[['lon', 'lat']] = sites['geometry'].str.extract(r'c\(([-\d.]+),\s*([-\d.]+)\)')
sites[['lon', 'lat']] = sites[['lon', 'lat']].astype(float)

sites
```

```python
sites.to_csv('data/processed/air_quality_nyc_sites.csv', index=False)
```

```python
# merge aqi with site locations
aqi = aqi.merge(sites[['aqs_id_full', 'site_name','lon', 'lat']], on='aqs_id_full', how='left')

aqi
```

```python
aqi = aqi[aqi['site_name'].notna()]

aqi = aqi.reset_index(drop=True)
```

```python
aqi
```

```python
# restrict to 2023 onwards
aqi['datetime'] = pd.to_datetime(aqi['datetime'])
aqi = aqi[aqi['datetime'] >= '2023-01-01']
```

```python
aqi['value'].describe()
```

```python
aqi
```

```python
index_cols = ['aqs_id_full', 'pollutant', 'unit', 'site_name', 'lon', 'lat']

aqi_daily = aqi.groupby(index_cols + [aqi['datetime'].dt.date])['value'].agg(['min', 'max']).reset_index()
aqi_daily.rename(columns={'datetime': 'date', 'min': 'daily_low', 'max': 'daily_high'}, inplace=True)
aqi_daily['date'] = pd.to_datetime(aqi_daily['date'])

aqi_daily
```

```python
# 90 day moving average for daily high
aqi_daily['date'] = pd.to_datetime(aqi_daily['date'])
aqi_daily = aqi_daily.sort_values(by=['site_name', 'date'])
aqi_daily['daily_high_90d_ma'] = aqi_daily.groupby('site_name')['daily_high'].transform(lambda x: x.rolling(window=90, min_periods=1).mean())

# 90 day moving average for daily low
aqi_daily['date'] = pd.to_datetime(aqi_daily['date'])
aqi_daily = aqi_daily.sort_values(by=['site_name', 'date'])
aqi_daily['daily_low_90d_ma'] = aqi_daily.groupby('site_name')['daily_low'].transform(lambda x: x.rolling(window=90, min_periods=1).mean())

aqi_daily
```

```python
# melt
aqi_daily_melted = aqi_daily.melt(id_vars=index_cols + ['date'], value_vars=['daily_low', 'daily_high', 'daily_low_90d_ma', 'daily_high_90d_ma'], var_name='measure', value_name='concentration')

aqi_daily_melted['date'] = pd.to_datetime(aqi_daily_melted['date'])

```

```python
aqi_daily_melted = aqi_daily_melted[aqi_daily_melted['site_name'].isin(
    ['Queens College','Midtown-DOT', 'BQE', 'Hamilton Bridge', 'Mott Haven', 'SI Expwy', 'Broadway/35th St']
)]
```

```python
aqi_daily_melted['site_name'] = aqi_daily_melted['site_name'].map({
    'Queens College': 'Queens College (Queens)',
    'Midtown-DOT': 'Midtown-DOT (Manhattan)',
    'BQE': 'BQE (Brooklyn)',
    'Hamilton Bridge': 'Hamilton Bridge (Bronx)',
    'Mott Haven': 'Mott Haven (Bronx)',
    'SI Expwy': 'SI Expwy (Staten Island)',
    'Broadway/35th St': 'Broadway/35th St (Manhattan)'
})
```

```python
# order sites by borough
borough_order = {
    'Midtown-DOT (Manhattan)': 1,
    'Broadway/35th St (Manhattan)': 2,
    'BQE (Brooklyn)': 3,
    'Queens College (Queens)': 4,
    'Hamilton Bridge (Bronx)': 5,
    'Mott Haven (Bronx)': 6,
    'SI Expwy (Staten Island)': 7
}

aqi_daily_melted['site_order'] = aqi_daily_melted['site_name'].map(borough_order)
aqi_daily_melted = aqi_daily_melted.sort_values(by=['site_order', 'date'])
aqi_daily_melted = aqi_daily_melted.drop(columns=['site_order'])
```

```python
aqi_daily_melted.to_csv('data/processed/air_quality_nyc_daily.csv', index=False)
```
