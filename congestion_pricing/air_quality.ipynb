```python
import pandas as pd
```

```python
air_quality_url = "https://raw.githubusercontent.com/timothyfraser/nyc_congestion_rep/refs/heads/main/data/aqi.csv"
```

```python
aqi = pd.read_csv(air_quality_url)

aqi.head()
```

```python
aqi['site_agency'].value_counts()
```

```python
aqi_nyc = aqi[aqi['site_agency'] == 'New York Dept. of Environmental Conservation']

aqi_nyc
```

```python
aqi_nyc['site_name'].unique()
```

```python
aqi_nyc = aqi_nyc[aqi_nyc['site_name'].isin([
    'Fresh Kills', 'Port Richmond', 'Bklyn - PS 314', 'CCNY', 'Manhattan/IS143', 'Bklyn - PS274',
    'Morrisania', 'Bronx - IS52', 'Maspeth', 'Bronx - IS74', 'Queens',
])]

aqi_nyc
```

```python
aqi_nyc['unit'].value_counts()
```

```python
aqi_nyc['datetime'] = pd.to_datetime(aqi_nyc['datetime'])

aqi_nyc
```

```python
# restrict to 2024 onwards
aqi_nyc = aqi_nyc[aqi_nyc['datetime'] >= '2023-05-31']

aqi_nyc
```

```python
index_cols = ['lat', 'lon', 'param', 'unit', 'site_name']

aqi_daily = aqi_nyc.groupby(index_cols + [aqi_nyc['datetime'].dt.date])['concentration'].agg(['min', 'max']).reset_index()
aqi_daily.rename(columns={'datetime': 'date', 'min': 'daily_low', 'max': 'daily_high'}, inplace=True)

aqi_daily
```

```python
# 90 day moving average for daily high
aqi_daily['date'] = pd.to_datetime(aqi_daily['date'])
aqi_daily = aqi_daily.sort_values(by=['site_name', 'date'])
aqi_daily['daily_high_90d_ma'] = aqi_daily.groupby('site_name')['daily_high'].transform(lambda x: x.rolling(window=90, min_periods=1).mean())
```

```python
# 90 day moving average for daily low
aqi_daily['date'] = pd.to_datetime(aqi_daily['date'])
aqi_daily = aqi_daily.sort_values(by=['site_name', 'date'])
aqi_daily['daily_low_90d_ma'] = aqi_daily.groupby('site_name')['daily_low'].transform(lambda x: x.rolling(window=90, min_periods=1).mean())
```

```python
# melt
aqi_daily_melted = aqi_daily.melt(id_vars=index_cols + ['date'], value_vars=['daily_low', 'daily_high', 'daily_low_90d_ma', 'daily_high_90d_ma'], var_name='measure', value_name='concentration')

aqi_daily_melted['date'] = pd.to_datetime(aqi_daily_melted['date'])
```

```python
import numpy as np

# replace -999.0 with NaN
aqi_daily_melted['concentration'] = aqi_daily_melted['concentration'].replace(-999.0, np.nan)
```

```python
aqi_daily_melted.to_csv('data/processed/air_quality_nyc_daily.csv', index=False)
```
