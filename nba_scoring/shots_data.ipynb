```python
from nba_api.stats.endpoints import shotchartleaguewide
import pandas as pd

def get_league_wide_shots(season='2025-26', timeout=30, retries=10):
    """
    Get all league shots in one API call using ShotChartLeagueWide
    """
    print(f"Fetching league-wide shot data for {season}...")
    time.sleep(1)  # be polite to the API
    try:
        shots = shotchartleaguewide.ShotChartLeagueWide(
            league_id='00',
            season=season
        )
        
        # Check how many dataframes are returned
        dfs = shots.get_data_frames()
        print(f"✓ Number of dataframes returned: {len(dfs)}")
        
        # Look at each dataframe
        for i, df in enumerate(dfs):
            print(f"\n--- DataFrame {i} ---")
            print(f"Rows: {len(df)}")
            print(f"Columns: {df.columns.tolist()}")
            print(f"\nFirst few rows:")
            print(df.head())
            print("\n")
        
        return dfs
        
    except Exception as e:
        print(f"✗ Error: {e}")
        import traceback
        traceback.print_exc()
        return None

```

```python
# make a list to hold data
season_shots_list = []

# get league wide shots for each season
for season in seasons:
    shots = get_league_wide_shots(season=season)
    time.sleep(1)  # brief pause between calls
    season_shots_list.append(shots)
```

```python
# robustly concatenate the results in season_shots_list (handles lists-of-dfs, single dfs, Nones)
def _iter_dfs(obj):
    """Yield DataFrame objects from nested lists/tuples or single DataFrame; ignore other types."""
    if obj is None:
        return
    if isinstance(obj, pd.DataFrame):
        yield obj
    elif isinstance(obj, (list, tuple)):
        for item in obj:
            yield from _iter_dfs(item)
    else:
        return

dfs = [df for item in season_shots_list for df in _iter_dfs(item)]

if dfs:
    shots = pd.concat(dfs, ignore_index=True)
else:
    shots = pd.DataFrame()

shots
```

```python
# Loop seasons, add SEASON column, consolidate zones per season and combine results

def consolidate_shot_zones(df_in):
    """
    Calculate FGA and FGM for each basic zone, consolidating corner 3s.
    Expects columns: SHOT_ZONE_BASIC, FGA, FGM
    """
    df_local = df_in.copy()
    df_local['ZONE_CONSOLIDATED'] = df_local['SHOT_ZONE_BASIC'].replace({
        'Left Corner 3': 'Corner 3',
        'Right Corner 3': 'Corner 3'
    })
    consolidated = df_local.groupby('ZONE_CONSOLIDATED', as_index=False).agg({
        'FGA': 'sum',
        'FGM': 'sum'
    })
    return consolidated

seasons = [f"{y}-{str(y+1)[-2:]}" for y in range(1996, 2026)]  # 1996-97 .. 2025-26

season_consolidated_list = []

for season in seasons:
    print(f"Fetching season {season} ...")
    res = get_league_wide_shots(season=season)
    if res is None:
        print(f"  → no data for {season}, skipping")
        continue

    # handle function returning list of dfs or a single df
    if isinstance(res, list):
        if len(res) == 0:
            print(f"  → empty result list for {season}, skipping")
            continue
        df_season = res[0].copy()
    elif isinstance(res, pd.DataFrame):
        df_season = res.copy()
    else:
        print(f"  → unexpected return type for {season}: {type(res)}, skipping")
        continue

    # Ensure FGA/FGM exist
    if not {'FGA', 'FGM', 'SHOT_ZONE_BASIC'}.issubset(df_season.columns):
        print(f"  → required columns missing for {season}, skipping")
        continue

    df_season['SEASON'] = season
    consolidated = consolidate_shot_zones(df_season)
    consolidated['SEASON'] = season
    season_consolidated_list.append(consolidated)

# Combine all seasons
if season_consolidated_list:
    df = pd.concat(season_consolidated_list, ignore_index=True)
    # Optional: sort for readability
    df = df.sort_values(['SEASON', 'FGA'], ascending=[True, False]).reset_index(drop=True)
    print(f"\nCombined consolidated zones for {len(seasons)} seasons. Rows: {len(df)}")
else:
    df = pd.DataFrame()
    print("No season data consolidated.")

# Show result
df
```

```python
# re-order columns for readability
df = df[['SEASON', 'ZONE_CONSOLIDATED', 'FGA', 'FGM']]

# rename columns
df.columns = ['Season', 'Zone', 'Attempts', 'Makes']

# calculate misses
df['Misses'] = df['Attempts'] - df['Makes']

# show result
df
```

```python
def pivot_zones_wide(df):
    """
    Convert long format to wide format with columns like:
    'Mid-Range Attempts', 'Mid-Range Makes', 'Mid-Range Misses', etc.
    """
    # Melt the numeric columns to create a long format first
    df_melted = df.melt(
        id_vars=['Season', 'Zone'],
        value_vars=['Attempts', 'Makes', 'Misses'],
        var_name='Metric',
        value_name='Value'
    )
    
    # Create new column names by combining Zone and Metric
    df_melted['Column'] = df_melted['Zone'] + ' ' + df_melted['Metric']
    
    # Pivot to wide format
    df_wide = df_melted.pivot(
        index='Season',
        columns='Column',
        values='Value'
    ).reset_index()
    
    # Sort columns alphabetically (optional)
    cols = ['Season'] + sorted([c for c in df_wide.columns if c != 'Season'])
    df_wide = df_wide[cols]
    
    return df_wide

# Apply the transformation
df_wide = pivot_zones_wide(df)

df_wide
```

```python
# restrict to only the season col and attempts cols
df_attempts_only = df_wide[['Season'] + [col for col in df_wide.columns if 'Attempts' in col]]

df_attempts_only.columns

# sort by distance
df_attempts_only = df_attempts_only[['Season', 'Restricted Area Attempts', 'In The Paint (Non-RA) Attempts',
                                   'Mid-Range Attempts', 'Corner 3 Attempts', 'Above the Break 3 Attempts', 'Backcourt Attempts']]

# rename columns for readability
df_attempts_only.columns = ['Season', 'Restricted Area', 'In The Paint (Non-RA)', 'Mid-Range', 'Corner 3', 'Above the Break 3', 'Backcourt']

df_attempts_only
```
