```python
import nfl_data_py as nfl
import pandas as pd

# Import team descriptive information
teams = nfl.import_team_desc()

# The 'teams' DataFrame will contain various columns related to NFL teams,
# including unique identifiers. Common ID columns you might find are:
# - 'team_id': A unique identifier for the team.
# - 'team_abbr': The common abbreviation for the team (e.g., 'BUF', 'NE').
# - 'team_name': The full name of the team (e.g., 'Buffalo Bills', 'New England Patriots').

# To view the available columns and the first few rows of the data:
print(teams.head())

# To specifically extract a list of team IDs:
team_ids = teams['team_id'].tolist()
print("\nList of NFL Team IDs:")
print(team_ids)

# save team names and ids to json
teams[['team_id', 'team_name']].to_json('data/nfl_teams.json', orient='records', indent=4)
```

```python
url = "https://raw.githubusercontent.com/djblechn-su/nba-player-team-ids/refs/heads/master/NBA_Team_IDs.csv"

# depulicate by "NBA_Current_Link_ID"
teams = pd.read_csv(url).drop_duplicates(subset=["NBA_Current_Link_ID"])

teams = teams[['BBRef_Team_Name', 'NBA_Current_Link_ID']]

teams.to_json('data/nba_teams.json', orient='records', indent=4)
```

```python
import statsapi
```

```python
# Get all MLB teams and their IDs
mlb_teams = statsapi.get('teams', {'sportIds': 1})['teams']

# Extract team names and IDs
team_data = [{'team_id': team['id'], 'team_name': team['name']} for team in mlb_teams]

# Convert to DataFrame
mlb_teams_df = pd.DataFrame(team_data)

# write to json
mlb_teams_df.to_json('data/mlb_teams.json', orient='records', indent=4)
```
